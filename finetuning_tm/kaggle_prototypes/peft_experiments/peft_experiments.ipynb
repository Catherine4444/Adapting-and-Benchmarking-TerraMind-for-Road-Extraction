# HPO: Rank

```python
import yaml
def setup_kaggle_cfg(data_cfg_path:str , cfg_path:str, wandb_project_name:str = None):
    with open(data_cfg_path) as f:
        cfg = yaml.safe_load(f)
        
    cfg["init_args"]["dataset_dir"] = "/kaggle/input/datasets/cath4444/rosadataset"
    cfg["init_args"]["num_workers"] = 4
    
    with open(data_cfg_path, "w") as f:
        yaml.safe_dump(cfg, f, sort_keys=False)

    if wandb_project_name:
        with open(cfg_path) as f:
            cfg = yaml.safe_load(f)
            
        cfg["trainer"]["logger"]["init_args"]["project"] = wandb_project_name
        
        with open(cfg_path, "w") as f:
            yaml.safe_dump(cfg, f, sort_keys=False)
    

def setup_rank_cfg(cfg_path:str, r:int):
    with open(cfg_path) as f:
        cfg = yaml.safe_load(f)

    cfg["model"]["init_args"]["model_args"]["peft_config"]["peft_config_kwargs"]["r"] = r
    
    with open(cfg_path, "w") as f:
        yaml.safe_dump(cfg, f, sort_keys=False)

def tune_lora_rank(name:str, seeds:list =[0,1,2], ranks:list=[4,8,16,32,64]):
    %cd /kaggle/working/InstaRoadPrototype/src/finetuning_tm
    data_cfg_path = "config/rosa_data/base_data.yml"
    
    cfg_path = f"config/peft/lora_linear_{name}_modules.yml"
    wandb_project_name = f"lora_linear_{name}_tune_rank_kaggle_runs_rosa"
    
    setup_kaggle_cfg(data_cfg_path, cfg_path, wandb_project_name)
    
    for seed in seeds:
        for rank in ranks:
            setup_rank_cfg(cfg_path, rank)
            
            ! terratorch fit \
                --config {cfg_path} \
                --trainer.max_epochs 20 \
                --seed_everything {seed} \
                --trainer.default_root_dir /kaggle/working/outputs/lora_{name}_seed{seed}_rank{rank}\
                --trainer.precision 16-mixed \
                --data {data_cfg_path} \
                --custom_modules_path /kaggle/working/InstaRoadPrototype/src/finetuning_tm
```

## Attn

```python
tune_lora_rank("attn")
```
